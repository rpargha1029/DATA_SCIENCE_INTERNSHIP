## 2. Objective
Build and evaluate Scikit-learn predictive models to predict passenger survival (`Survived`) based on demographic and ticket features.

## 3. Dataset Description
- **Source**: [Kaggle Titanic Dataset](https://www.kaggle.com/datasets/yasserh/titanic-dataset)

## 4. Library Imports

```python
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.append(str(Path('..').resolve() / 'src'))
from train_evaluate import locate_dataset, load_and_clean_data, train_and_evaluate
```

## 5. Dataset Loading

```python
csv_path = locate_dataset()
raw_df = pd.read_csv(csv_path)
print('Dataset shape:', raw_df.shape)
raw_df.head()
```

## 6. Data Understanding

```python
print(raw_df.info())
print('\nMissing values:\n', raw_df.isnull().sum())
raw_df.describe()
```

## 7. Data Cleaning

```python
df = load_and_clean_data(csv_path)
print('Cleaned modeling dataframe shape:', df.shape)
```

## 8. Implementation (Model Training & Evaluation)

```python
summary = train_and_evaluate(df, save_artifacts=True)
print('Best Model:', summary['best_model'])
```

## 9. Visualizations

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.countplot(data=df, x='Sex', hue='Survived', ax=axes[0])
axes[0].set_title('Survival by Sex')
sns.countplot(data=df, x='Pclass', hue='Survived', ax=axes[1])
axes[1].set_title('Survival by Passenger Class')
plt.tight_layout()
plt.show()
```

## 10. Results

```python
metrics_table = pd.DataFrame({
    name: {k: v for k, v in m.items() if k in ['accuracy', 'precision', 'recall', 'f1_score', 'mse', 'r2_score']}
    for name, m in summary['models'].items()
}).T
metrics_table
```
