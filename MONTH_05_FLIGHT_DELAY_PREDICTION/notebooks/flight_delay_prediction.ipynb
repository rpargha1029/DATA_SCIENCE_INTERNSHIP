## 1. Project Title
**Flight Delay Prediction Using Machine Learning**

## 2. Objective
Train Logistic Regression and Random Forest classifiers to predict whether a flight will be delayed or on time using airline, route, and schedule features.

## 3. Dataset Description
- **Source**: Airline Delay Dataset (Kaggle)
- **Expected Local Path**: `../data/<airline_delay>.csv`

## 4. Library Imports

```python
import sys
from pathlib import Path
import pandas as pd

sys.path.append(str(Path('..').resolve() / 'src'))
from train_flight_delay import locate_dataset, prepare_flight_data, train_and_evaluate
```

## 5. Dataset Loading & 6. Data Understanding

```python
csv_path = locate_dataset()
raw_df = pd.read_csv(csv_path, low_memory=False)
print(raw_df.shape)
raw_df.head()
```

## 7. Data Cleaning & Leakage-Free Feature Preparation

```python
X, y, num_cols, cat_cols = prepare_flight_data(raw_df)
print('Numeric:', num_cols, '| Categorical:', cat_cols)
```

## 8. Implementation (Logistic Regression & Random Forest)

```python
summary = train_and_evaluate(X, y, num_cols, cat_cols, save_artifacts=True)
print('Best Model:', summary['best_model'])
```

## 9. Visualizations & 10. Results

```python
for name, m in summary['models'].items():
    print(f'=== {name} ===')
    print(m['classification_report'])
```
