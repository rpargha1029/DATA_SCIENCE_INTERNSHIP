# Month 4: Content-Based Movie Recommendation System
## 1. Project Title
**Build a Content-Based Movie Recommendation System (TMDB 5000)**

## 2. Objective
Build a content-based recommender system using movie descriptions, genres, keywords, and cast with TF-IDF vectorization and cosine similarity.

## 3. Dataset Description
- **Source**: [TMDB 5000 Movie Dataset (Kaggle)](https://www.kaggle.com/datasets/tmdb/tmdb-movie-metadata)
- **Expected Local Paths**: `../data/tmdb_5000_movies.csv` and `../data/tmdb_5000_credits.csv`

## 4. Library Imports

```python
import sys
from pathlib import Path
import pandas as pd

sys.path.append(str(Path('..').resolve() / 'src'))
from recommender import locate_datasets, prepare_movie_tags, build_and_save_model, recommend
```

## 5. Dataset Loading & 6. Data Understanding

```python
movies_path, credits_path = locate_datasets()
movies_df = pd.read_csv(movies_path)
credits_df = pd.read_csv(credits_path)
print('Movies:', movies_df.shape, '| Credits:', credits_df.shape)
```

## 7. Data Cleaning (Combine tags, lowercase, remove punctuation)

```python
movies_clean = prepare_movie_tags(movies_df, credits_df)
movies_clean.head()
```

## 8. Implementation (TF-IDF Vectorization & Cosine Similarity)

```python
vectorizer, similarity = build_and_save_model(movies_clean, save_artifacts=True)
print('Similarity matrix shape:', similarity.shape)
```

## 9. Visualizations & 10. Results

```python
recommend('Avatar', movies_clean, similarity, top_n=5)
```

## 11. Interpretation & 12. Conclusion
The recommender groups movies sharing thematic keywords, genres, and cast members, and persists the vectorizer and similarity matrix as `.pkl` artifacts in `../models/`.
